```python
# ==========================================================
# AI-POWERED CUSTOMER EMOTION INTELLIGENCE SYSTEM
# FINAL ERROR-FREE VERSION
# ==========================================================

import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns

from wordcloud import WordCloud
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import nltk

from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# =========================
# DOWNLOAD NLTK DATA
# =========================

nltk.download("punkt")
nltk.download("stopwords")

# =========================
# SAMPLE DATASET (NO EXTERNAL FILE NEEDED)
# =========================

data = {
    "Text": [
        "This product is amazing and works perfectly",
        "Worst purchase ever, very bad quality",
        "I am very happy with the service",
        "Delivery was late and product was damaged",
        "Excellent quality and fast delivery",
        "Totally disappointed with this product",
        "Great experience, will buy again",
        "Not worth the money",
        "Customer support was very helpful",
        "Terrible experience, I want a refund"
    ] * 300  # expand dataset automatically
}

df = pd.DataFrame(data)

print("Dataset loaded:", len(df))

# =========================
# CLEAN TEXT
# =========================

stop_words = set(stopwords.words("english"))

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z ]", "", text)
    words = word_tokenize(text)
    return " ".join([w for w in words if w not in stop_words])

df["clean"] = df["Text"].apply(clean_text)

# =========================
# SENTIMENT ANALYSIS
# =========================

analyzer = SentimentIntensityAnalyzer()

def sentiment(text):
    score = analyzer.polarity_scores(text)["compound"]
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"

df["sentiment"] = df["clean"].apply(sentiment)

# =========================
# SIMPLE EMOTION DETECTION (RULE-BASED)
# =========================

def emotion(text):
    text = text.lower()

    if any(w in text for w in ["happy", "good", "great", "excellent", "amazing"]):
        return "Joy"
    elif any(w in text for w in ["bad", "worst", "terrible", "refund", "disappointed"]):
        return "Anger"
    elif "late" in text or "damaged" in text:
        return "Sadness"
    elif "helpful" in text:
        return "Trust"
    else:
        return "Neutral"

df["emotion"] = df["Text"].apply(emotion)

# =========================
# CUSTOMER SCORE
# =========================

df["score"] = df["sentiment"].map({
    "Positive": 100,
    "Neutral": 50,
    "Negative": 10
})

# =========================
# CHURN RISK
# =========================

def churn(x):
    if x < 30:
        return "High Risk"
    elif x < 70:
        return "Medium Risk"
    return "Low Risk"

df["churn"] = df["score"].apply(churn)

# =========================
# BUSINESS REPORT
# =========================

print("\n================ REPORT ================")
print("Total Reviews:", len(df))
print("\nSentiment Distribution:\n", df["sentiment"].value_counts())
print("\nEmotion Distribution:\n", df["emotion"].value_counts())
print("\nChurn Risk:\n", df["churn"].value_counts())

# =========================
# VISUALIZATION 1
# =========================

plt.figure(figsize=(6,4))
sns.countplot(x=df["sentiment"])
plt.title("Sentiment Analysis")
plt.show()

# =========================
# VISUALIZATION 2
# =========================

plt.figure(figsize=(7,4))
sns.countplot(y=df["emotion"], order=df["emotion"].value_counts().index)
plt.title("Emotion Analysis")
plt.show()

# =========================
# WORDCLOUD
# =========================

text = " ".join(df[df["sentiment"]=="Positive"]["clean"])

wc = WordCloud(width=800, height=400, background_color="white").generate(text)

plt.figure(figsize=(10,5))
plt.imshow(wc)
plt.axis("off")
plt.title("Positive Reviews WordCloud")
plt.show()

# =========================
# FINAL OUTPUT
# =========================

df.to_csv("final_emotion_output.csv", index=False)

print("\nPROJECT COMPLETED SUCCESSFULLY ✔")
print("File saved as final_emotion_output.csv")
```
