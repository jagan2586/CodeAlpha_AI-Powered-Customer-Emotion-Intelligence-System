```python
#=================================================
 #AI-POWERED CUSTOMER EMOTION INTELLIGENCE SYSTEM
#==================================================
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# =========================
# SIMPLE INBUILT DATASET
# =========================

data = {
    "text": [
        "I love this product",
        "This is amazing",
        "Worst experience ever",
        "Very bad quality",
        "I am happy with service",
        "Totally disappointed",
        "Excellent product",
        "Not good at all",
        "Great experience",
        "I want refund",
        "Very satisfied customer",
        "Terrible service"
    ] * 200
}

df = pd.DataFrame(data)

print("Dataset Loaded:", len(df))

# =========================
# SIMPLE SENTIMENT LOGIC
# =========================

def sentiment(text):
    text = text.lower()

    positive_words = ["love", "amazing", "happy", "excellent", "great", "satisfied"]
    negative_words = ["worst", "bad", "terrible", "disappointed", "refund", "not"]

    if any(word in text for word in positive_words):
        return "Positive"
    elif any(word in text for word in negative_words):
        return "Negative"
    else:
        return "Neutral"

df["sentiment"] = df["text"].apply(sentiment)

# =========================
# SIMPLE EMOTION DETECTION
# =========================

def emotion(text):
    text = text.lower()

    if "love" in text or "happy" in text or "great" in text:
        return "Joy"
    elif "bad" in text or "worst" in text or "terrible" in text:
        return "Anger"
    elif "disappointed" in text or "refund" in text:
        return "Sadness"
    else:
        return "Neutral"

df["emotion"] = df["text"].apply(emotion)

# =========================
# CUSTOMER SCORE
# =========================

def score(sentiment):
    if sentiment == "Positive":
        return 100
    elif sentiment == "Neutral":
        return 50
    else:
        return 10

df["score"] = df["sentiment"].apply(score)

# =========================
# CHURN RISK
# =========================

def churn(x):
    if x < 30:
        return "High Risk"
    elif x < 70:
        return "Medium Risk"
    return "Low Risk"

df["churn"] = df["score"].apply(churn)

# =========================
# OUTPUT
# =========================

print("\n=== SENTIMENT COUNT ===")
print(df["sentiment"].value_counts())

print("\n=== EMOTION COUNT ===")
print(df["emotion"].value_counts())

print("\n=== CHURN RISK ===")
print(df["churn"].value_counts())

# =========================
# VISUALIZATION
# =========================

plt.figure(figsize=(6,4))
df["sentiment"].value_counts().plot(kind="bar")
plt.title("Sentiment Analysis")
plt.show()

plt.figure(figsize=(6,4))
df["emotion"].value_counts().plot(kind="bar")
plt.title("Emotion Analysis")
plt.show()

# =========================
# FINAL SAVE
# =========================

df.to_csv("final_output.csv", index=False)

print("\nPROJECT RUN SUCCESSFULLY ✔")
print("Output saved as final_output.csv")
```
